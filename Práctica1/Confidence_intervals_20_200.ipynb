# Intervalos de Confianza — Práctica 1: Simulación Ad Hoc

Este notebook calcula, para cada una de las 5 medidas de desempeño, la media muestral, la desviación estándar muestral, la media anchura (*half-width*) y los intervalos de confianza al 95% y al 99%, siguiendo el procedimiento del **Ejemplo 2** de la sección 1.11.1 del Capítulo 1 de [Banks1998].

Se aplica a los dos escenarios de la práctica:
- **Simulación con 20 clientes** (10 corridas / *runs*)
- **Simulación con 200 clientes** (10 corridas / *runs*)

```python
import numpy as np
from scipy import stats
from IPython.display import display, Markdown
import pandas as pd
```

## Función de cálculo de intervalo de confianza

Implementa las ecuaciones (2), (3) y (4) del Ejemplo 2 de [Banks1998]:

$$\bar{X} = \frac{\sum_{i=1}^{n} X_i}{n} \qquad S^2 = \frac{\sum_{i=1}^{n} X_i^2 - n\bar{X}^2}{n-1} \qquad h = t_{n-1,\,1-\alpha/2}\frac{S}{\sqrt{n}}$$

```python
def confidence_interval(data, alpha, label=""):
    """Calcula media, desviacion estandar muestral, half-width e IC para 'data' con nivel alpha.
    Muestra los resultados en formato Markdown, igual que el Ejemplo 2 de [Banks1998]."""
    reps = len(data)
    confidence = (1 - alpha) * 100

    mean = np.mean(data)
    var = np.var(data, ddof=1)
    desv = np.sqrt(var)

    tval = stats.t.ppf(1 - (alpha / 2), reps - 1)
    hval = tval * (desv / np.sqrt(reps))

    display(Markdown(
        rf"""
**{label}** — ${confidence:.1f}\%$ confidence

$\bar{{X}} = {mean:.4f}$

$S = {desv:.4f}$

$t_{{n-1,1-\frac{{\alpha}}{{2}}}} = t_{{{reps-1},{1-alpha/2}}} = {tval:.4f}$

$h = {hval:.4f}$

Intervalo de confianza : $C.I. = (\bar{{X}}-h , \bar{{X}}+h) = ({mean-hval:.4f}, {mean+hval:.4f})$

---
"""))

    return {
        "label": label,
        "confidence": confidence,
        "mean": mean,
        "S": desv,
        "t": tval,
        "h": hval,
        "ci_low": mean - hval,
        "ci_high": mean + hval,
    }
```

## 3. Tabla resumen comparativa (20 vs 200 clientes)

Se compara la desviación estándar muestral $S$ y el rango del intervalo de confianza ($2h$) para cada medida de desempeño, en ambos escenarios y niveles de confianza, con el fin de responder: **¿cuál combinación de número de clientes, repeticiones y porcentaje de confianza tiene la menor desviación estándar de la muestra y el menor rango del intervalo de confianza?**

```python
resultados = [
    res_ats_20_95, res_ats_20_99, res_ats_200_95, res_ats_200_99,
    res_pit_20_95, res_pit_20_99, res_pit_200_95, res_pit_200_99,
    res_awc_20_95, res_awc_20_99, res_awc_200_95, res_awc_200_99,
    res_fhw_20_95, res_fhw_20_99, res_fhw_200_95, res_fhw_200_99,
    res_aww_20_95, res_aww_20_99, res_aww_200_95, res_aww_200_99,
]

resumen = pd.DataFrame(resultados)
resumen["rango_IC"] = resumen["ci_high"] - resumen["ci_low"]
resumen = resumen[["label", "confidence", "mean", "S", "h", "ci_low", "ci_high", "rango_IC"]]
resumen.columns = ["Medida", "% Confianza", "Media", "S", "half-width (h)", "IC inf.", "IC sup.", "Rango IC"]
resumen.round(4)
```

```python
# Fila con el menor S y menor rango de IC en toda la tabla
print("Menor desviación estándar (S):")
display(resumen.loc[[resumen["S"].idxmin()]].round(4))

print("\nMenor rango de intervalo de confianza:")
display(resumen.loc[[resumen["Rango IC"].idxmin()]].round(4))
```
